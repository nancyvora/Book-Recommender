```python
from langchain_community.document_loaders import TextLoader
from langchain_text_splitters import  CharacterTextSplitter
from langchain_openai import OpenAIEmbeddings
from langchain_chroma import Chroma
import time
```

```python
from dotenv import load_dotenv

path = "../.env"

load_dotenv(dotenv_path=path, override=True)
```

```python
import pandas as pd

books = pd.read_csv("../data/processed/books_cleaned.csv")
```

```python
books["tagged_description"].to_csv("../data/processed/tagged_description.txt", index=False, header=False)
```

```python
raw_documents = TextLoader("../data/processed/tagged_description.txt",encoding="utf-8").load()
```

```python
text_splitter = CharacterTextSplitter(chunk_size=1, chunk_overlap=0, separator="\n")
documents = text_splitter.split_documents(raw_documents)
```

```python
documents[0]
```

```python
embeddings = OpenAIEmbeddings(model="text-embedding-3-small", chunk_size=100)

db_books = Chroma(
    embedding_function=embeddings,
    persist_directory="../data/chroma_db"
)

batch_size = 300

for i in range(0, len(documents), batch_size):
    batch = documents[i : i + batch_size]
    db_books.add_documents(batch)
    print(f"Added {i + len(batch)} / {len(documents)}")
    time.sleep(60)   # wait a minute so the limit resets
```

```python
query = "A book to teach children about nature"
docs = db_books.similarity_search(query, k=10)
docs
```

```python
books[books['isbn13'] == int(docs[0].page_content.split()[0].strip().strip('"'))]
```

```python
def retrive_semantic_recommendations(
        query: str,
        top_k: int = 10,
) -> pd.DataFrame:
    recs = db_books.similarity_search(query, k = 50)

    books_list = []

    for i in range(0, len(recs)):
        books_list += [int(recs[i].page_content.strip('"').split()[0])]

    return books[books["isbn13"].isin(books_list)].head(top_k)
```

```python
retrive_semantic_recommendations("A book to teach children about nature")
```
